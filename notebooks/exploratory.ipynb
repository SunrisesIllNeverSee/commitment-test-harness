# Exploratory Data Analysis and Visualization

This notebook is used for exploratory data analysis and visualization of the results from the commitment test harness. It integrates HuggingFace pipelines for processing linguistic signals and utilizes spaCy for extracting hard commitments.

```python
import pandas as pd
import matplotlib.pyplot as plt
from src.harness import run_tests
from src.extraction import extract_hard_commitments
from src.metrics import jaccard_index
from src.plotting import plot_fidelity

# Set parameters for the test harness
compression_thresholds = [10, 20, 30, 40, 50]
results = []

# Run tests for each compression threshold
for threshold in compression_thresholds:
    fidelity = run_tests(threshold)
    results.append(fidelity)

# Convert results to DataFrame for analysis
results_df = pd.DataFrame(results, columns=['Threshold', 'Fidelity'])

# Display the results
results_df.head()
```

```python
# Plot the fidelity results
plt.figure(figsize=(10, 6))
plot_fidelity(results_df['Threshold'], results_df['Fidelity'])
plt.title('Fidelity vs Compression Threshold')
plt.xlabel('Compression Threshold')
plt.ylabel('Fidelity')
plt.grid(True)
plt.show()
```
